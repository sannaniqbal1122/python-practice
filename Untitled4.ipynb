```python
import re
def extract_age(age):
  age_num = re.findall('[0-9]+', str(age))
  if len(age_num)>0:
    return age_num [0]
  else:
    return age
df['Age'] = df['Age'].apply(lambda x: extract_age(x) )
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
```

```python
df_age = df[df['Age'] != 'nan yeras']['Age']
```

```python
df_age
```

```python
age_median = int(df_age.dropna().astype('int64').median())
```

```python
age_median
```

```python
df.replace('nan year', age_median, inplace=True)
```

```python
df['Age'].fillna( age_median, inplace=True )
```
